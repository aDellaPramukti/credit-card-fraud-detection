```python
# =============================================================================
# 1. IMPORT LIBRARY
# =============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve,
    precision_recall_curve,
    average_precision_score
)

import warnings
warnings.filterwarnings('ignore')

print("Library berhasil di-import.")
# =============================================================================
# 2. UPLOAD DATASET
# =============================================================================

uploaded = files.upload()

df = pd.read_csv('creditcard.csv')

print("Dataset berhasil dibaca.")

# =============================================================================
# 3. DATASET OVERVIEW
# =============================================================================

print("Jumlah baris dan kolom:")
print(df.shape)

print("\n5 data pertama:")
display(df.head())

print("\nInformasi dataset:")
df.info()

# =============================================================================
# 4. CEK MISSING VALUE
# =============================================================================

missing_values = df.isnull().sum()

print("Jumlah missing value setiap kolom:")
display(missing_values[missing_values > 0])

if missing_values.sum() == 0:
    print("Dataset tidak memiliki missing value.")

# =============================================================================
# 5. STATISTIK DESKRIPTIF
# =============================================================================

display(df.describe())

# =============================================================================
# 6. DISTRIBUSI CLASS
# =============================================================================

class_counts = df['Class'].value_counts()

print("Jumlah transaksi berdasarkan Class:")
print(class_counts)

print("\nPersentase transaksi:")
print(df['Class'].value_counts(normalize=True) * 100)

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Class')

plt.title('Distribusi Transaksi Normal dan Fraud')
plt.xlabel('Class (0 = Normal, 1 = Fraud)')
plt.ylabel('Jumlah Transaksi')

plt.show()

# =============================================================================
# 7. PERSENTASE CLASS
# =============================================================================

class_percentage = df['Class'].value_counts(normalize=True) * 100

plt.figure(figsize=(7, 5))

plt.pie(
    class_percentage,
    labels=['Normal', 'Fraud'],
    autopct='%1.3f%%',
    startangle=90
)

plt.title('Persentase Transaksi Normal dan Fraud')

plt.show()

# =============================================================================
# 8. KORELASI FITUR TERHADAP TARGET
# =============================================================================

correlation = df.corr()['Class'].sort_values(ascending=False)

print("Korelasi fitur terhadap Class:")
display(correlation)

plt.figure(figsize=(8, 10))

correlation.drop('Class').sort_values().plot(kind='barh')

plt.title('Korelasi Fitur terhadap Class')
plt.xlabel('Correlation')
plt.ylabel('Feature')

plt.show()

# =============================================================================
# 9. MEMISAHKAN FEATURE DAN TARGET
# =============================================================================

X = df.drop('Class', axis=1)
y = df['Class']

print("Jumlah feature:", X.shape[1])
print("Jumlah data:", X.shape[0])

# =============================================================================
# 10. TRAIN TEST SPLIT
# =============================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Jumlah data training :", X_train.shape[0])
print("Jumlah data testing  :", X_test.shape[0])

print("\nDistribusi Class pada data training:")
print(y_train.value_counts())

print("\nDistribusi Class pada data testing:")
print(y_test.value_counts())

# =============================================================================
# 11. FEATURE SCALING
# =============================================================================

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

columns_to_scale = ['Time', 'Amount']

X_train_scaled[columns_to_scale] = scaler.fit_transform(
    X_train[columns_to_scale]
)

X_test_scaled[columns_to_scale] = scaler.transform(
    X_test[columns_to_scale]
)

print("Scaling berhasil dilakukan.")


# =============================================================================
# 12. LOGISTIC REGRESSION
# =============================================================================

logistic_model = LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)

logistic_model.fit(
    X_train_scaled,
    y_train
)

print("Logistic Regression selesai dilatih.")

# =============================================================================
# 13. PREDIKSI LOGISTIC REGRESSION
# =============================================================================

y_pred_lr = logistic_model.predict(X_test_scaled)

y_prob_lr = logistic_model.predict_proba(X_test_scaled)[:, 1]

print("Prediksi selesai.")

# =============================================================================
# 14. EVALUASI LOGISTIC REGRESSION
# =============================================================================

accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr = precision_score(y_test, y_pred_lr)
recall_lr = recall_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr)
roc_auc_lr = roc_auc_score(y_test, y_prob_lr)
average_precision_lr = average_precision_score(y_test, y_prob_lr)

print("=== LOGISTIC REGRESSION ===")

print(f"Accuracy          : {accuracy_lr:.4f}")
print(f"Precision         : {precision_lr:.4f}")
print(f"Recall            : {recall_lr:.4f}")
print(f"F1-Score          : {f1_lr:.4f}")
print(f"ROC-AUC           : {roc_auc_lr:.4f}")
print(f"Average Precision : {average_precision_lr:.4f}")

print("Classification Report - Logistic Regression")
print(
    classification_report(
        y_test,
        y_pred_lr,
        target_names=['Normal', 'Fraud']
    )
)

# =============================================================================
# 16. CONFUSION MATRIX
# =============================================================================

cm_lr = confusion_matrix(y_test, y_pred_lr)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Normal', 'Fraud'],
    yticklabels=['Normal', 'Fraud']
)

plt.title('Confusion Matrix - Logistic Regression')
plt.xlabel('Predicted')
plt.ylabel('Actual')

plt.show()

# =============================================================================
# 17. RANDOM FOREST
# =============================================================================

random_forest_model = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(
    X_train,
    y_train
)

print("Random Forest selesai dilatih.")

# =============================================================================
# 18. PREDIKSI RANDOM FOREST
# =============================================================================

y_pred_rf = random_forest_model.predict(X_test)

y_prob_rf = random_forest_model.predict_proba(X_test)[:, 1]

print("Prediksi selesai.")

# =============================================================================
# 19. EVALUASI RANDOM FOREST
# =============================================================================

accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)
roc_auc_rf = roc_auc_score(y_test, y_prob_rf)
average_precision_rf = average_precision_score(y_test, y_prob_rf)

print("=== RANDOM FOREST ===")

print(f"Accuracy          : {accuracy_rf:.4f}")
print(f"Precision         : {precision_rf:.4f}")
print(f"Recall            : {recall_rf:.4f}")
print(f"F1-Score          : {f1_rf:.4f}")
print(f"ROC-AUC           : {roc_auc_rf:.4f}")
print(f"Average Precision : {average_precision_rf:.4f}")

print("Classification Report - Random Forest")

print(
    classification_report(
        y_test,
        y_pred_rf,
        target_names=['Normal', 'Fraud']
    )
)

# =============================================================================
# 21. CONFUSION MATRIX RANDOM FOREST
# =============================================================================

cm_rf = confusion_matrix(y_test, y_pred_rf)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_rf,
    annot=True,
    fmt='d',
    cmap='Greens',
    xticklabels=['Normal', 'Fraud'],
    yticklabels=['Normal', 'Fraud']
)

plt.title('Confusion Matrix - Random Forest')
plt.xlabel('Predicted')
plt.ylabel('Actual')

plt.show()

# =============================================================================
# 22. PERBANDINGAN MODEL
# =============================================================================

comparison = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Random Forest'
    ],
    'Accuracy': [
        accuracy_lr,
        accuracy_rf
    ],
    'Precision': [
        precision_lr,
        precision_rf
    ],
    'Recall': [
        recall_lr,
        recall_rf
    ],
    'F1-Score': [
        f1_lr,
        f1_rf
    ],
    'ROC-AUC': [
        roc_auc_lr,
        roc_auc_rf
    ],
    'Average Precision': [
        average_precision_lr,
        average_precision_rf
    ]
})

display(comparison.round(4))

# =============================================================================
# 23. VISUALISASI PERBANDINGAN MODEL
# =============================================================================

comparison_plot = comparison.set_index('Model')

comparison_plot[
    ['Precision', 'Recall', 'F1-Score', 'ROC-AUC']
].plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Perbandingan Performa Model')
plt.xlabel('Model')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.ylim(0, 1)

plt.legend()
plt.show()

# =============================================================================
# 24. ROC CURVE
# =============================================================================

fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr_lr,
    tpr_lr,
    label=f'Logistic Regression (AUC = {roc_auc_lr:.4f})'
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label=f'Random Forest (AUC = {roc_auc_rf:.4f})'
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    label='Random Classifier'
)

plt.title('ROC Curve')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.legend()

plt.show()

# =============================================================================
# 25. PRECISION-RECALL CURVE
# =============================================================================

precision_lr_curve, recall_lr_curve, _ = precision_recall_curve(
    y_test,
    y_prob_lr
)

precision_rf_curve, recall_rf_curve, _ = precision_recall_curve(
    y_test,
    y_prob_rf
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall_lr_curve,
    precision_lr_curve,
    label='Logistic Regression'
)

plt.plot(
    recall_rf_curve,
    precision_rf_curve,
    label='Random Forest'
)

plt.title('Precision-Recall Curve')
plt.xlabel('Recall')
plt.ylabel('Precision')

plt.legend()

plt.show()

# =============================================================================
# 26. FEATURE IMPORTANCE
# =============================================================================

feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': random_forest_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

display(feature_importance.head(15))

plt.figure(figsize=(10, 7))

top_features = feature_importance.head(15)

sns.barplot(
    data=top_features,
    x='Importance',
    y='Feature'
)

plt.title('Top 15 Feature Importance - Random Forest')
plt.xlabel('Importance')
plt.ylabel('Feature')

plt.show()


# =============================================================================
# 27. KESIMPULAN HASIL MODEL
# =============================================================================

print("=" * 70)
print("KESIMPULAN")
print("=" * 70)

print(f"""
Dataset memiliki {len(df):,} transaksi dengan {df['Class'].sum():,}
transaksi yang dikategorikan sebagai fraud.

Model yang digunakan:
1. Logistic Regression
2. Random Forest

Logistic Regression:
- Accuracy          : {accuracy_lr:.4f}
- Precision         : {precision_lr:.4f}
- Recall            : {recall_lr:.4f}
- F1-Score          : {f1_lr:.4f}
- ROC-AUC           : {roc_auc_lr:.4f}
- Average Precision : {average_precision_lr:.4f}

Random Forest:
- Accuracy          : {accuracy_rf:.4f}
- Precision         : {precision_rf:.4f}
- Recall            : {recall_rf:.4f}
- F1-Score          : {f1_rf:.4f}
- ROC-AUC           : {roc_auc_rf:.4f}
- Average Precision : {average_precision_rf:.4f}
""")

```
